<b> Crossover graph

```python
telephone_global = df1.iloc[:,4:].sum()
mobile_global = df2.iloc[:,4:].sum()
broadband_global = df3.iloc[:,4:].sum()

telephone_global.index = telephone_global.index.astype(int)
mobile_global.index = mobile_global.index.astype(int)
broadband_global.index = broadband_global.index.astype(int)
```

```python
import pandas as pd

df_cross = pd.DataFrame({
    "Year": telephone_global.index,
    "Telephone": telephone_global.values,
    "Mobile": mobile_global.values,
    "Broadband": broadband_global.values
})

# Mobile surpassing Telephone
mobile_cross = df_cross[df_cross["Mobile"] > df_cross["Telephone"]].iloc[0]["Year"]

# Broadband surpassing Telephone
bb_cross = df_cross[df_cross["Broadband"] > df_cross["Telephone"]].iloc[0]["Year"]

print("Mobile surpassed Telephone in:", mobile_cross)
print("Broadband surpassed Telephone in:", bb_cross)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

plt.plot(df_cross["Year"], df_cross["Telephone"], label="Telephone")
plt.plot(df_cross["Year"], df_cross["Mobile"], label="Mobile")
plt.plot(df_cross["Year"], df_cross["Broadband"], label="Broadband")

plt.axvline(x=mobile_cross, linestyle="--", label="Mobile > Telephone")
plt.axvline(x=bb_cross, linestyle="--", label="Broadband > Telephone")

plt.title("Communication Technology Crossover Graph")
plt.xlabel("Year")
plt.ylabel("Global Subscriptions")
plt.legend()

plt.grid(True)
plt.show()
```
